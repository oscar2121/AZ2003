```python
# ==============================================================================
# 1. IMPORTACIÓN DE LIBRERÍAS
# ==============================================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Conjunto de datos y preprocesamiento
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Modelos de Machine Learning Supervisado
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

# Métricas de evaluación
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Configuración visual
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
print("Librerías importadas exitosamente.")
```

```python
# ==============================================================================
# 2. CARGA Y PREPROCESAMIENTO DE DATOS
# ==============================================================================
cancer_data = load_breast_cancer()
X = pd.DataFrame(cancer_data.data, columns=cancer_data.feature_names)
y = pd.Series(cancer_data.target, name="target") # 0 = Maligno, 1 = Benigno (ajustado en análisis)

print(f"Dimensiones del dataset: {X.shape[0]} muestras x {X.shape[1]} atributos")
print("Distribución de clases:")
print(f" - Clase 0 (Maligno): {(y == 0).sum()} muestras")
print(f" - Clase 1 (Benigno): {(y == 1).sum()} muestras\n")

# Para que el análisis coincida con la clínica (1 = Maligno = clase de interés):
y_clinico = (y == 0).astype(int)

# Partición Estratificada en Train (80%) y Test (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y_clinico, test_size=0.20, random_state=42, stratify=y_clinico
)

# Estandarización de características con StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Conjunto de entrenamiento: {X_train.shape[0]} observaciones")
print(f"Conjunto de prueba: {X_test.shape[0]} observaciones")
```

```python
# ==============================================================================
# 3. DEFINICIÓN Y ENTRENAMIENTO DE LOS 6 ALGORITMOS
# ==============================================================================
modelos = {
    "Regresión Logística": LogisticRegression(random_state=42, max_iter=1000),
    "Árbol de Decisión": DecisionTreeClassifier(random_state=42, max_depth=5),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Support Vector Machine (SVM)": SVC(kernel='rbf', probability=True, random_state=42),
    "K-Nearest Neighbors (KNN)": KNeighborsClassifier(n_neighbors=5),
    "Naive Bayes (Gaussiano)": GaussianNB()
}

resultados = []
predicciones = {}

for nombre, modelo in modelos.items():
    # Entrenar con datos escalados
    modelo.fit(X_train_scaled, y_train)
    y_pred = modelo.predict(X_test_scaled)
    predicciones[nombre] = y_pred

    # Calcular métricas
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    # Matriz de confusión
    cm = confusion_matrix(y_test, y_pred)
    vn, fp, fn, vp = cm.ravel()

    resultados.append({
        "Algoritmo": nombre,
        "Accuracy": round(acc, 4),
        "Precision": round(prec, 4),
        "Recall": round(rec, 4),
        "F1-Score": round(f1, 4),
        "Falsos Positivos (FP)": fp,
        "Falsos Negativos (FN)": fn
    })

# Convertir a DataFrame ordenado por F1-Score
df_resultados = pd.DataFrame(resultados).sort_values(by="F1-Score", ascending=False)
print("=== TABLA COMPARATIVA DE RESULTADOS ===")
display(df_resultados)
```

```python
# ==============================================================================
# 4. VISUALIZACIÓN DE MATRICES DE CONFUSIÓN
# ==============================================================================
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

etiquetas = ['Benigno (0)', 'Maligno (1)']

for idx, (nombre, y_pred) in enumerate(predicciones.items()):
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[idx],
                xticklabels=etiquetas, yticklabels=etiquetas, cbar=False)
    axes[idx].set_title(f"{nombre}\nF1-Score: {f1_score(y_test, y_pred):.4f}", fontsize=12, fontweight='bold')
    axes[idx].set_xlabel('Predicción del Modelo')
    axes[idx].set_ylabel('Clase Real')

plt.suptitle('Comparación de Matrices de Confusión en Test (n = 114)', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# 5. GRÁFICO COMPARATIVO DE MÉTRICAS (ACCURACY, PRECISION, RECALL, F1)
# ==============================================================================
metricas_plot = df_resultados.melt(
    id_vars=['Algoritmo'],
    value_vars=['Accuracy', 'Precision', 'Recall', 'F1-Score'],
    var_name='Métrica', value_name='Valor'
)

plt.figure(figsize=(14, 6))
chart = sns.barplot(data=metricas_plot, x='Algoritmo', y='Valor', hue='Métrica', palette='Set2')
plt.title('Rendimiento Comparativo de Modelos de Clasificación Supervisada', fontsize=14, fontweight='bold')
plt.ylim(0.80, 1.02)
plt.ylabel('Puntaje (0.0 a 1.0)')
plt.xticks(rotation=20, ha='right')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
